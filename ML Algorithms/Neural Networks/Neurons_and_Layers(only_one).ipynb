# One neurons and layers

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras import Sequential
from tensorflow.keras.losses import MeanSquaredError, BinaryCrossentropy
from tensorflow.keras.activations import sigmoid
```

```python
x_train = np.array([[1.0], [2.0]], dtype=np.float32)
y_train = np.array([[300.0], [500.0]], dtype=np.float32)
```

## Neuron without activation - Regression/Linear Model

```python
linear_layer=tf.keras.layers.Dense(units=1,activation='linear')
```

```python
linear_layer.get_weights() # empty bcz we have not yet instantiated
```

```python
a1 = linear_layer(x_train[0].reshape(1,1)) # we must pass input as a 2-D array compulsory
print(a1)
```

```python
w, b= linear_layer.get_weights() # randomly assigned the weights
print(f'w = {w}, b={b}')
```

```python
set_w = np.array([[200]])
set_b = np.array([100])
linear_layer.set_weights([set_w, set_b])
```

```python
a1 = linear_layer(x_train[0].reshape(1,1))
print(a1)
alin = np.dot(set_w,x_train[0].reshape(1,1)) + set_b
print(alin)
```
