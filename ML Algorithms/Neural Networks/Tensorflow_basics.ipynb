## changing the shape

```python
tensor1=tf.ones([3]) #1-d array with size 3
tensor2=tf.ones([1,3]) #one 2-d array containing one 1-d array of size 3
tensor3=tf.ones([1,2,3]) #one 2-d array containing two 1-d array each of size 3
print(tensor3)
```

```python
tensor4=tf.reshape(tensor3,[2,3,1]) #tot ele=2*3*1=6 so this number must be equal with the tot elements in the tensor3
print(tensor4)
```

```python
tensor5=tf.reshape(tensor3,[2,3])
print(tensor5)
```

```python
tensor6=tf.reshape(tensor3,[3,-1])
print(tensor6)
```
