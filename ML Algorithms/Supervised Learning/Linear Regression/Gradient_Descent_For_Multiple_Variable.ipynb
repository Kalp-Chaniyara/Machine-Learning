# Gradient Descent for multiple variable using numpy

```python
import math,copy
import numpy as np
import matplotlib.pyplot as plt
```

# Problem Statement

    | Size (sqft) | Number of Bedrooms  | Number of floors | Age of  Home | Price (1000s dollars)  |
    | ----------------| ------------------- |----------------- |--------------|-------------- |
    | 2104            | 5                   | 1                | 45           | 460           |
    | 1416            | 3                   | 2                | 40           | 232           |
    | 852             | 2                   | 1                | 35           | 178           |

```python
x_train = np.array([[2104, 5, 1, 45], [1416, 3, 2, 40], [852, 2, 1, 35]])
y_train = np.array([460, 232, 178])
```

```python
def compute_cost(x,y,w,b):
    m=x.shape[0]
    n=x.shape[1]
    cost=0
    for i in range(m):
        f_wb=np.dot(x[i],w)+b
        cost+=(f_wb-y[i])**2
    tot_cost=cost/2*m

    return tot_cost
```

```python
def compute_gradient(x,y,w,b):
    m=x.shape[0]
    n=x.shape[1]
    dj_dw=np.zeros((n,))
    dj_db=0
    for i in range(m):
        f_wb=(np.dot(x[i],w)+b)-y[i]
        for j in range(n):
            dj_dw[j]+=(f_wb)*x[i,j]
        dj_db+=f_wb
    dj_dw/=m
    dj_db/=m

    return dj_dw,dj_db
```

```python
#whole iteration
def gradient_descent(x,y,w_in,b_in,alpha,num_iters,compute_cost,compute_gradient):
    w=copy.deepcopy(w_in)
    b=b_in
    m=x.shape[0]
    n=x.shape[1]

    for i in range(num_iters):
        dj_dw,dj_db=compute_gradient(x,y,w,b)
        w=w-alpha*dj_dw
        b=b-alpha*dj_db

        cost=compute_cost(x,y,w,b)

        if i%100==0:
            print(f"iteration {i:4}, cost {cost:5}, dj_dw {dj_dw}, dj_db {dj_db:5}, w {w}, b {b:5}")

    return w,b
```

```python
ini_w=np.zeros((4,))
ini_b=0
num_iters=1000
alpha=5.0e-7
w_fin,b_fin=gradient_descent(x_train,y_train,ini_w,ini_b,alpha,num_iters,compute_cost,compute_gradient)

print(f"this is final w {w_fin}, b {b_fin}")
```

```python
m=x_train.shape[0]
for i in range(m):
    print(f"final prediction is {np.dot(w_fin,x_train[i])+b_fin}, target value is {y_train[i]}")
```
