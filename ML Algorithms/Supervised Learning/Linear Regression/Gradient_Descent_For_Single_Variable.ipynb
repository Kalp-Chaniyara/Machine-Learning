# Gradient Descent for single variable using numpy

```python
import math,copy
import numpy as np
import matplotlib.pyplot as plt
```

# Problem Statement

    | Size (1000 sqft)     | Price (1000s of dollars) |
    | ----------------| ------------------------ |
    | 1               | 300                      |
    | 2               | 500                      |

```python
x_train=np.array([1.0,2.0])
y_train=np.array([300.0,500.0])
```

```python
#compute cost
def compute_cost(x,y,w,b):
    m=x.shape[0]
    cost=0
    for i in range(m):
        f_wb=w*x[i]+b
        cost+=(f_wb-y[i])**2
    total_cost=cost/2*m

    return total_cost
```

```python
#derivative
def compute_gradient(x,y,w,b):
    m=x.shape[0]
    dj_dw=0
    dj_db=0

    for i in range(m):
        f_wb=w*x[i]+b
        dj_dw_i=(f_wb-y[i])*x[i]
        dj_db_i=f_wb-y[i]
        dj_dw+=dj_dw_i
        dj_db+=dj_db_i
    dj_dw=dj_dw/m
    dj_db=dj_db/m

    return dj_dw,dj_db
```

```python
def gradient_descent(x,y,w_in,b_in,alpha,num_iters,cost_fun,grad_fun):
    # w=copy.deepcopy(w_in)
    w=w_in
    b=b_in

    for i in range(num_iters):
        dj_dw,dj_db=grad_fun(x,y,w,b)
        w=w-alpha*dj_dw
        b=b-alpha*dj_db
        cost=cost_fun(x,y,w,b)

        if i%1000==0:
            print(f"iteration {i:4}, cost {cost:5}, dj_dw {dj_dw:5}, dj_db {dj_db:5}, w {w:5}, b {b:5}")
    return w,b
```

```python
w_in=0
b_in=0
ite=10000
alpha=1.0e-2
w_fin,b_fin=gradient_descent(x_train,y_train,w_in,b_in,alpha,ite,compute_cost,compute_gradient)

print(f"this is the final w {w_fin}, b {b_fin}")
# print(w_in)
```

```python
print(f"1000 sqft house prediction is {w_fin*1.0 + b_fin:0.1f} Thousand dollars")
print(f"1200 sqft house prediction is {w_fin*1.2 + b_fin:0.1f} Thousand dollars")
print(f"2000 sqft house prediction is {w_fin*2.0 + b_fin:0.1f} Thousand dollars")
```
