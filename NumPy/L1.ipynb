# indexing, slicing and iteration

```python
arr12=np.arange(24)
arr12
```

```python
arr12.reshape(6,4)
```

```python
arr12[2:] #all same as the list
```

```python
arr13=np.arange(24).reshape(6,4)
arr13
```

```python
arr13[:,2] #2nd column of all the rows
```

```python
arr13[:,1:3] #1st and 2nd columns of all the rows
```

```python
for i in np.nditer(arr13):
    print(i)
```

# numpy operations

```python
arr14=np.array([1,2,3,4,5,6])
arr15=np.array([7,8,9,10,11,12])
```

```python
arr14-arr15
```

```python
arr14*arr15
```

```python
arr14*2
```

```python
arr14>3
```

```python
arr16=np.arange(6).reshape(2,3)
arr17=np.arange(6,18).reshape(3,4)
```

```python
arr18=arr16.dot(arr17) #simple matrix multiplication
arr18
```

```python
arr18.max()
```

```python
arr18.min()
```

```python
arr18.max(axis=0) #give me maximum elements columns wise
```

```python
arr18.max(axis=1) #give me maximum elements row wise
```

# Reshaping numpy array

```python
arr18.ravel() #convert higher dimentional array to 1-d array
```

```python
arr18.transpose()
```
