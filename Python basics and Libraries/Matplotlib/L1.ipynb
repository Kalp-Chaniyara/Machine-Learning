# Scatter Plot

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5,6,7]
y=[2,3,1,4,5,3,6]
plt.scatter(x,y)
# plt.scatter(x,y,c='r')
plt.title('Scatter Plot')
plt.xlabel('day')
plt.ylabel('number')
plt.show()
```
